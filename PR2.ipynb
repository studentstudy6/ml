```python

from collections import deque

# Graph representation
graph = {
    'S': ['A', 'B'],
    'A': ['C'],
    'B': ['G'],
    'C': ['D'],
    'D': ['G'],
    'G': []
}

# Function to reconstruct the path
def get_path(parent, start, goal):
    path = []
    current = goal

    while current is not None:
        path.append(current)
        current = parent[current]

    path.reverse()
    return path


# Breadth First Search
def bfs(start, goal):
    queue = deque([start])
    visited = {start}
    parent = {start: None}

    while queue:
        node = queue.popleft()

        if node == goal:
            return get_path(parent, start, goal)

        for neighbour in graph[node]:
            if neighbour not in visited:
                visited.add(neighbour)
                parent[neighbour] = node
                queue.append(neighbour)

    return None


# Depth First Search
def dfs(start, goal):
    stack = [start]
    visited = {start}
    parent = {start: None}

    while stack:
        node = stack.pop()

        if node == goal:
            return get_path(parent, start, goal)

        # Reverse to visit neighbours in their listed order
        for neighbour in reversed(graph[node]):
            if neighbour not in visited:
                visited.add(neighbour)
                parent[neighbour] = node
                stack.append(neighbour)

    return None


# Main program
start = 'S'
goal = 'G'

print("UNINFORMED SEARCH ALGORITHMS")
print("----------------------------")

bfs_path = bfs(start, goal)
print("BFS Path:", " -> ".join(bfs_path))

dfs_path = dfs(start, goal)
print("DFS Path:", " -> ".join(dfs_path))
```
