```python

import random
import time

print("SMART HOME INTELLIGENT AGENT")
print("Agent started automatically...")
print("--------------------------------")

try:
    while True:

        # Simulate automatic sensor readings
        temperature = random.randint(20, 40)
        motion_detected = random.choice([True, False])
        light_intensity = random.randint(0, 100)
        door_open = random.choice([True, False])
        owner_present = random.choice([True, False])

        print("\n--- SENSOR READINGS ---")
        print("Temperature:", temperature, "°C")
        print("Motion Detected:", motion_detected)
        print("Light Intensity:", light_intensity)
        print("Door Open:", door_open)
        print("Owner Present:", owner_present)

        print("\n--- AGENT ACTIONS ---")

        # Automatic temperature control
        if temperature > 30:
            print("AC: ON")
        else:
            print("AC: OFF")

        # Automatic light control
        if motion_detected and light_intensity < 40:
            print("Light: ON")
        else:
            print("Light: OFF")

        # Automatic security control
        if door_open and not owner_present:
            print("Security Alarm: ON")
        else:
            print("Security Alarm: OFF")

        print("--------------------------------")

        # Wait before next sensor reading
        time.sleep(3)

except KeyboardInterrupt:
    print("\nSmart Home Agent stopped.")
```
